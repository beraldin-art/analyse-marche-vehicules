#### PHASE B : Modélisation (Conception)

Dans l’architecture Delta Lake (Databricks), on organise les données en trois niveaux : 
- Bronze : Données brutes directement importées
- Silver : Données nettoyées, normalisées
- Golden : Données prêtes pour l'analyse ou le reporting

Nous vons donc fait une modélisation en étoile, en créant une table de fait et 4 tables de dimensions (date, vehicule, localisation et vendeur) lesquelles ont été alimentées à partir de la table silver. 

Le modèle en étoile que nous avons mis en place permet de structurer les données de manière simple, lisible et optimisée pour faciliter l'exploitation et l’analyse. La séparation entre la table de faits (contenant les mesures comme le prix ou le kilométrage) et 4 tables de dimensions (véhicule, date, localisation, vendeur) facilite les jointures, améliore les performances des requêtes analytiques et rend le modèle compatible avec les outils de reporting, Power BI en particulier. 


```python
%sql
/* TRAITEMENT DES DONNEESS : Création de la table de silver */

CREATE OR REPLACE VIEW silver_annonces AS
SELECT
  CAST(ID AS BIGINT) AS id_annonce,
  -- Dates
  TO_DATE(Date, 'yyyy-MM-dd') AS date_annonce,
  
  -- Prix & kilométrage
  CAST(prix AS INT) AS prix,
  CAST(mileage AS INT) AS kilometrage,
  
  -- Motorisation
  CASE fuel
    WHEN '1' THEN 'Essence'
    WHEN '2' THEN 'Diesel'
    WHEN '4' THEN 'Electrique'
    ELSE 'Inconnu'
  END AS motorisation,
  
  -- Boîte de vitesse
  CASE gearbox
    WHEN '1' THEN 'Manuelle'
    WHEN '2' THEN 'Automatique'
    ELSE 'Inconnu'
  END AS type_boite,
  
  -- Vendeur
  CASE Vendeur
    WHEN 'private' THEN 'Particulier'
    WHEN 'pro' THEN 'Professionnel'
    ELSE 'Inconnu'
  END AS type_vendeur,
  
  -- Géographie
  regexp_extract(cp, '(\\d{5})', 1) AS code_postal,
  TRIM(ville) AS ville,
  
  -- Autres colonnes
  brand,
  model,
  regdate,
  nvl(vehicle_type, 'Autre') AS type_vehicule
FROM bronze_annonces
WHERE Prix IS NOT NULL;
```

```python
%sql
/* Visualisation de la table silver*/
select * from silver_annonces
```

```python
%sql
/* CREATION DES TABLES DE DIMENSIONS : VEHICULE */
CREATE OR REPLACE TABLE dim_vehicule AS
SELECT
  CONCAT('veh', ROW_NUMBER() OVER (
      ORDER BY marque, modele, motorisation, type_vehicule, type_boite
  )) AS id_vehicule,
  marque,
  modele,
  motorisation,
  type_vehicule,
  type_boite
FROM (
  SELECT DISTINCT
    brand AS marque,
    model AS modele,
    motorisation,
    type_vehicule,
    type_boite
  FROM silver_annonces
) t;


```

```python
%sql
/* CREATION DES TABLES DE DIMENSIONS : DATE */
CREATE OR REPLACE TABLE dim_date AS
SELECT
  CONCAT('dat', ROW_NUMBER() OVER (ORDER BY date_annonce)) AS id_date,
  date_annonce AS date,
  YEAR(date_annonce) AS annee,
  MONTH(date_annonce) AS mois,
  DAY(date_annonce) AS jour,
  DATE_FORMAT(date_annonce, 'EEEE') AS jour_semaine
FROM (
  SELECT DISTINCT
    date_annonce
  FROM silver_annonces
  WHERE date_annonce IS NOT NULL
) d;
```

```python
%sql
/* CREATION DES TABLES DE DIMENSIONS : LOCALISATION */

CREATE OR REPLACE TABLE dim_localisation AS
SELECT
  CONCAT('loc', ROW_NUMBER() OVER (ORDER BY code_postal, ville)) AS id_localisation,
  code_postal,
  ville
FROM (
  SELECT DISTINCT
    code_postal,
    ville
  FROM silver_annonces
  WHERE code_postal IS NOT NULL AND ville IS NOT NULL
) l;

```

```python
%sql
/* CREATION DES TABLES DE DIMENSIONS : VENDEUR */

CREATE OR REPLACE TABLE dim_vendeur AS
SELECT
  CONCAT('ven', ROW_NUMBER() OVER (ORDER BY type_vendeur)) AS id_vendeur,
  type_vendeur
FROM (
  SELECT DISTINCT
    type_vendeur
  FROM silver_annonces
  WHERE type_vendeur IS NOT NULL
) v;

```

```python
%sql
/* CREATION DE LA TABLE DE FAIT */

CREATE OR REPLACE TABLE fact_annonce AS
SELECT DISTINCT
    s.id_annonce,
    d.id_date,
    v.id_vehicule,
    l.id_localisation,
    ven.id_vendeur,
    s.prix,
    s.kilometrage
FROM silver_annonces s
LEFT JOIN dim_date d
    ON s.date_annonce = d.date
LEFT JOIN dim_vehicule v
    ON s.brand          = v.marque
   AND s.model          = v.modele
   AND s.motorisation   = v.motorisation
   AND s.type_vehicule  = v.type_vehicule
   AND s.type_boite     = v.type_boite
LEFT JOIN dim_localisation l
    ON s.code_postal = l.code_postal
   AND s.ville       = l.ville
LEFT JOIN dim_vendeur ven
    ON s.type_vendeur = ven.type_vendeur
WHERE s.id_annonce IS NOT NULL;
```

Après avoir chargé les différentes tables, nous remarquons que la table de fait comporte 738 observatios au lieu de 739 que comporte la table silver. Pour identifier le problème, nous avons recherché les identifiants des annonces en double. Nous avons ainsi consaté qu'un identifiant de l'annonce était en double, et ces doublants comporte les même informations sur tous les attributs de la base de donnée. Donc la table de fait est bien correcte avec 738 observations

```python
%sql
/* Détection des doublons sur les identifiants */

SELECT id_annonce, COUNT(*) as valeur_double
FROM silver_annonces
GROUP BY id_annonce
HAVING COUNT(*) > 1;
```

#### Vérification des prix dérisoirs

En faisant une analyse descriptive du montant des véhicules, on remarque une plage des prix allant de 169 à 8 000 avec une moyenne de 5 044.96. Sur cette base, il se peut qu'il y ait des prix erronés. Nous faisons une vérification sur un seuil de 500. 

```python
%sql
SELECT min(f.prix) AS prix_min, round(avg(f.prix),2) AS prix_moyen, max(f.prix) AS prix_max from fact_annonce f 
```

```python
%sql
/* Vérification des prix dérisoires */

SELECT *
FROM fact_annonce f
JOIN dim_vehicule v ON f.id_vehicule = v.id_vehicule
WHERE f.prix <= 500; 
```
